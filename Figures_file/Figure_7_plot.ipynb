```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams

# ============================================================
# LOAD DATA
# ============================================================
data = np.load(r"Figure_7_data.npz")

eta_vals = data['eta_vals']
Nop_ss   = data['Nop_ss']
Nm       = data['Nm']

phi_density   = data['phi_density']
density_eta30 = data['density_eta30']

phi     = data['phi']
density = data['density']

# ============================================================
# STYLE
# ============================================================
rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.size": 14,
    "axes.labelsize": 16,
    "axes.titlesize": 16,
    "legend.fontsize": 10,
    "xtick.labelsize": 14,
    "ytick.labelsize": 14,
    "axes.linewidth": 0.8,
    "xtick.direction": "out",
    "ytick.direction": "out",
})

# ============================================================
# FIGURE
# ============================================================
fig, axes = plt.subplots(1, 2, figsize=(7.2, 3.0))

# (a)
ax = axes[0]
ax.plot(eta_vals, Nop_ss, lw=1.5, label=r'GP')
ax.plot(eta_vals, Nm, '--', lw=1.5, label=r'3-mode')
ax.axvline(x=30, ls=':', lw=1.2, color='k')
ax.set_xlabel(r'$\eta/\omega_r$')
ax.set_ylabel(r'$|\mathcal{N}|$')
ax.set_xlim(0, 100)
ax.set_ylim(0, 1)
ax.set_title(r'(a)', x=0.1, y=0.85)
ax.legend(frameon=False)

# (b)
ax = axes[1]
ax.plot(phi_density/(2*np.pi), density_eta30, lw=1.5, label=r'GP')
ax.plot(phi/(2*np.pi), density, '--', lw=1.3, label=r'3-mode')
ax.set_xlabel(r'$\phi/(2\pi)$')
ax.set_ylabel(r'$|\Psi(\phi)|^2/N$')
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_title(r'(b)', x=0.1, y=0.85)
ax.legend(frameon=False)

plt.tight_layout(pad=0.6)
plt.show()
```
