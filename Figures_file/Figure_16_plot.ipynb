```python
import numpy as np
import matplotlib.pyplot as plt
## Reload the data

data = np.load(r"Figure_16_data.npz")

f1 = data["f1"]
Sap_1 = data["Sap_1"]
Sbm_1 = data["Sbm_1"]

f1b = data["f1b"]
Sap1_1 = data["Sap1_1"]
Sbm1_1 = data["Sbm1_1"]

f2 = data["f2"]
Sap_2 = data["Sap_2"]
Sbm_2 = data["Sbm_2"]

f2b = data["f2b"]
Sap1_2 = data["Sap1_2"]
Sbm1_2 = data["Sbm1_2"]

## Plotting 

# ===============================
# LaTeX style
# ===============================
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.size": 18,
    "axes.labelsize": 18,
    "xtick.labelsize": 16,
    "ytick.labelsize": 16
})

# ==================================================
# Plot
# ==================================================
fig, ax = plt.subplots(2,4, figsize=(12,5))

# ---------- Row 1 : l1=8 l2=4 ----------
ax[0,0].plot(f1, Sap_1)
ax[0,0].set_xlim([0,10])
ax[0,0].set_ylabel(r"$S_{\alpha_{+}}$")
ax[0,0].text(-0.35,0.9,"(a)",transform=ax[0,0].transAxes)
ax[0,0].axvline(6.25, color='g', ls='--', lw=1.2)
ax[0,0].axvline(8.02, color='r', ls='--', lw=1.2)

ax[0,1].plot(f1, Sbm_1)
ax[0,1].set_xlim([0,10])
ax[0,1].set_ylabel(r"$S_{\beta_{-}}$")
ax[0,1].text(-0.35,0.9,"(b)",transform=ax[0,1].transAxes)
ax[0,1].axvline(1.36, color='g', ls='--', lw=1.2)
ax[0,1].axvline(2.25, color='r', ls='--', lw=1.2)

ax[0,2].plot(f1b, Sap1_1)
ax[0,2].set_xlim([0,50])
ax[0,2].set_ylabel(r"$S_{\alpha_{+}}$")
ax[0,2].text(-0.35,0.9,"(c)",transform=ax[0,2].transAxes)
ax[0,2].axvline(6.25, color='g', ls='--', lw=1.2)
ax[0,2].axvline(8.02, color='r', ls='--', lw=1.2)

ax[0,3].plot(f1b, Sbm1_1)
ax[0,3].set_xlim([0,50])
ax[0,3].set_ylabel(r"$S_{\beta_{-}}$")
ax[0,3].text(-0.35,0.9,"(d)",transform=ax[0,3].transAxes)
ax[0,3].axvline(1.36, color='g', ls='--', lw=1.2)
ax[0,3].axvline(2.25, color='r', ls='--', lw=1.2)

# ---------- Row 2 : l1=4 l2=8 ----------
ax[1,0].plot(f2, Sap_2)
ax[1,0].set_xlim([0,10])
ax[1,0].set_ylabel(r"$S_{\alpha_{+}}$")
ax[1,0].set_xlabel(r"$\omega/\omega_r$")
ax[1,0].text(-0.35,0.9,"(e)",transform=ax[1,0].transAxes)
ax[1,0].axvline(1.36, color='g', ls='--', lw=1.2)
ax[1,0].axvline(2.25, color='r', ls='--', lw=1.2)

ax[1,1].plot(f2, Sbm_2)
ax[1,1].set_xlim([0,10])
ax[1,1].set_ylabel(r"$S_{\beta_{-}}$")
ax[1,1].set_xlabel(r"$\omega/\omega_r$")
ax[1,1].text(-0.35,0.9,"(f)",transform=ax[1,1].transAxes)
ax[1,1].axvline(6.25, color='g', ls='--', lw=1.2)
ax[1,1].axvline(8.02, color='r', ls='--', lw=1.2)

ax[1,2].plot(f2b, Sap1_2)
ax[1,2].set_xlim([0,50])
ax[1,2].set_ylabel(r"$S_{\alpha_{+}}$")
ax[1,2].set_xlabel(r"$\omega/\omega_r$")
ax[1,2].text(-0.35,0.9,"(g)",transform=ax[1,2].transAxes)
ax[1,2].axvline(1.36, color='g', ls='--', lw=1.2)
ax[1,2].axvline(2.25, color='r', ls='--', lw=1.2)

ax[1,3].plot(f2b, Sbm1_2)
ax[1,3].set_xlim([0,50])
ax[1,3].set_ylabel(r"$S_{\beta_{-}}$")
ax[1,3].set_xlabel(r"$\omega/\omega_r$")
ax[1,3].text(-0.35,0.9,"(h)",transform=ax[1,3].transAxes)
ax[1,3].axvline(6.25, color='g', ls='--', lw=1.2)
ax[1,3].axvline(8.02, color='r', ls='--', lw=1.2)

# Paper-style ticks
for a in ax.flat:
    a.tick_params(direction="in")

plt.tight_layout()
plt.show()
```
